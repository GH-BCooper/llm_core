and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants
MODEL_GPT = 'gemini-2.5-flash'
MODEL_LLAMA = 'gemma4'
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('GEMINI_API_KEY')
if api_key:
    print("Nice api key broooo")
else:
    print("Bad api broooo")
```

```python
# For Gemini
gemini = OpenAI(
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)
```

```python
# For Ollama
ollama = OpenAI(
    base_url="http://localhost:11434/v1", 
    api_key="ollama"
)
```

```python
# here is the question; type over this to ask something new

question = """
What is system.out.println(x + y)?}
"""
question
```

```python
system_prompt = """
You are given a question of technical nature. You have to translate that question in easy to understand language and then explain it in simple words.
You are supposed to answer like:
The answer to your question: "question" is answer: "answer"
"""
system_prompt
```

```python
def get_user_prompt(question):
    user_prompt = f"""
Can you please explain this technical question to me.
The question is: {question}
Please explain in simple words with easy to understand  examples.
"""
    return user_prompt
```

```python
print(get_user_prompt(question))
```

```python
# Get gpt-4o-mini to answer, with streaming
def explain_question(question):
    stream = gemini.chat.completions.create(
        model=MODEL_GPT,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": get_user_prompt(question)}
          ],
        stream=True
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
explain_question(question)
```

```python
# Get Llama 3.2 to answer
def explain_question(question):
    stream = ollama.chat.completions.create(
        model=MODEL_LLAMA,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": get_user_prompt(question)}
          ],
        stream=True
    )    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```
